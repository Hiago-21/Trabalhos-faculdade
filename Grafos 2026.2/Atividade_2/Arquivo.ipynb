```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()
```

```python
G.add_node("Graveto"     )
G.add_node("Açucar"      )
G.add_node("Madeira"     )
G.add_node("Vidro"       )
G.add_node("Teia"        )
G.add_node("Ferro"       )
G.add_node("Redstone"    )
G.add_node("Corante"     )
G.add_node("OlhoDeAranha")
G.add_node("Cogumelo"    )
```

```python
G.add_edge("Vidro", "Vidro"  )
G.add_edge("Vidro", "Corante")

G.add_edge("Ferro", "Ferro"   )

G.add_edge("Açucar", "Cogumelo")

G.add_edge("Graveto", "Graveto" )
G.add_edge("Graveto", "Madeira" )
G.add_edge("Graveto", "Ferro"   )
G.add_edge("Graveto", "Redstone")
G.add_edge("Graveto", "Teia"    )

G.add_edge("Madeira", "Madeira" )
G.add_edge("Madeira", "Redstone")
G.add_edge("Madeira", "Ferro"   )
G.add_edge("Madeira", "Teia"    )

G.add_edge("Redstone", "Redstone")
G.add_edge("Redstone", "Ferro"   )

G.add_edge("OlhoDeAranha", "Açucar"  )
G.add_edge("OlhoDeAranha", "Cogumelo")
```

```python
plt.figure(figsize=(10, 8))
posicao = nx.circular_layout(G)
nx.draw(G, pos=posicao, with_labels=True, node_color='lightgreen', node_size=2500, font_weight='bold', font_size=9)

plt.show()
```

Fazendo o Gráfo complementar **COM** a função da biblioteca:
(A função serve apenas para grafos simples, então ela não pega as arestas dos vértices que ligam eles mesmos)

```python
G_complementar = nx.complement(G)

plt.figure(figsize=(10, 8))

posicao = nx.circular_layout(G_complementar)

nx.draw(G_complementar, pos=posicao, with_labels=True, node_color='lightcoral', node_size=2500, font_weight='bold', font_size=9)

plt.show()
```

Fazendo o Gráfo complementar **SEM** a função da biblioteca:

```python
G_comp = nx.Graph()

vertices = list(G.nodes())
G_comp.add_nodes_from(vertices)


for i in range(len(vertices)):
    for j in range(i, len(vertices)):
        item_A = vertices[i]
        item_B = vertices[j]
        
        if not G.has_edge(item_A, item_B):
            G_comp.add_edge(item_A, item_B)

plt.figure(figsize=(10, 8))
posicao = nx.circular_layout(G_comp)
nx.draw(G_comp, pos=posicao, with_labels=True, node_color='lightcoral', node_size=2500, font_weight='bold', font_size=9)
plt.show()
```
